```python
#Renaming Columns
#df2.rename(columns={'New Cases':'New_Cases','New Deaths':'New_Deaths',})
df2.columns=df2.columns.str.replace('_','')
df2.columns
```

```python
#stripping whitespace
df2['Country/Region']=df2['Country/Region'].str.strip()
df2['WHORegion']=df2['WHORegion'].str.strip()
```

```python
#locating records with '0' and inspecting for relevance
df2.loc[(df2 == 0).any(axis=1)]
```

```python
#Variances of numeric columns
print(df2.var(numeric_only=True))
```

```python
#Modes of numeric columns
modes = df2.mode(numeric_only=True)
modes
```

```python
#correlation matrix
corr_matrix= df2.corr(numeric_only=True)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(14, 10))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", linewidths=0.5)
plt.title("COVID-19 Metrics Correlation Matrix", fontsize=16 )
plt.show()
```

```python
# Skewness of all numeric columns
df2.skew(numeric_only=True)
```

```python
# Visualizations
numeric_cols = df.select_dtypes(include=['float64','int64']).columns
plt.figure(figsize=(20, 15))
for i, col in enumerate(numeric_cols, 1):
    plt.subplot((len(numeric_cols) // 3) + 1, 3, i)
    sns.histplot(df[col], kde=True, bins=30)
    plt.title(f"{col} (skew={df[col].skew():.2f})")
plt.tight_layout()
plt.show()
```

```python
#visualizing distribution of confirmed cases in countries using rolling mean
df2['Confirmed'].rolling(window=7).mean().plot()
plt.title('Distribution')
plt.show()
```

```python
#visualizing distribution of deaths in countries using rolling mean
df2['Deaths'].rolling(window=7).mean().plot()
plt.title('Distribution')
plt.show()
```

```python
#visualizing distribution of recovered cases in countries using rolling mean
df2['Recovered'].rolling(window=7).mean().plot()
plt.title('Distribution')
plt.show()
```

```python
# WHO region distribution
df2['WHORegion'].value_counts()
```

```python
#grouping by WHO region and calculating average confirmed cases
grp=df2.groupby('WHORegion')['Confirmed'].mean().sort_values(ascending=False)

#bar plot of average confirmed cases by WHO region
sns.barplot(x='WHORegion', y='Confirmed', data=df2, estimator='mean')
plt.xticks(rotation=45)
plt.title('Average Confirmed Cases by WHO Region')
plt.show()
```

```python
# Confirmes cases vs Deaths per 100 Cases
df2.sort_values(by='Deaths/100Cases', ascending=False)[['Country/Region', 'Deaths/100Cases']].head(10)
sns.scatterplot(x='Confirmed', y='Deaths/100Cases', data=df2)
plt.title('Confirmed Cases vs Deaths per 100 Cases')
plt.show()
```

```python
#v confirmed last week vs weekly change
df2.sort_values(by='1week%increase', ascending=False)[['Country/Region', '1week%increase']].head(10)
sns.scatterplot(x='Confirmedlastweek', y='1weekchange', data=df2)
plt.title('Confirmed Last Week vs Weekly Change')
plt.show()
```
