```python
#cleaning
#renaming columns
df2.rename(columns= {'Sales_Order #':'Sales_Order'}, inplace=True)
```

```python
#delete unnecessary columns
df2.drop(columns = ['Product_Category','Sub_Category','Month',], inplace = True)
df2.drop(columns = ['Year'], inplace = True)
```

```python
#filling null values with previous value
df2['Day'].fillna(method='ffill', inplace = True)
```

```python
#stripping whitespaces
df2.columns=df2.columns.str.strip()
```

```python
#using loc to strip
df2.loc[:, 'Age_Group'] = df2['Age_Group'].str.strip()
```

```python
#change data type
df2['Day']=df2['Day'].astype (str)
```

```python
df2.dtypes
```

```python
#locating and filling null values
df2.loc[df2['Product_Description'].isnull()]
df2['Product_Description']=df2['Product_Description'].fillna('Unknown')
```

```python
df2.loc[df2['Age_Group'].isnull()]
df2['Age_Group']=df2['Age_Group'].fillna('Adults (35-64)')
```

```python
df2.loc[df2['Order_Quantity'].isnull()]
df2['Order_Quantity'].dropna()
```

```python
df2
```

```python
#dropping records
df2.loc[df2['Cost']==0]
df2 = df2[df2['Cost'] != 0]
```

```python
#Handling Outliers; description, 
Max=[50,68,92,110, 79,62]
Q1=Max.quartile(0.25)
Q3=Max.quartile(0.75)
IQR=Q3-Q1
Lower=Q1-1.5*IQR
Upper=Q3+1.15*IQR
Outliers=Max[(Max<Lower)|(Max>Upper)]
print(Outliers)
```
